```python
total = 0
i = 0
while i < 10:
    #目前累積(6) + 這次要加(4) ->設定回去
    total = total + (i + 1)
    print(total)
    i = i + 1

    #2 / 7 -> 0.2857142857
    a, b = 2, 7
    ans = "0."
    i = 0
    while i < 10:
        #補0, x10, 2->20
        a10 = a * 10
        #取商(//) 20//7
        ans = ans + str(a10 // b)
        print(ans)
        #取餘(%) 20%7
        a = a10 % b
        i = i + 1


x, y = 0, 0 #U:上 D:下 L:左 R:右
commands = "ULLRDL" #x = -2 y = 0
#s = "hello"
#s[0] ->'h'
#s[1] ->'e'...

i = 0
#算長度使用len(s)
while i < len(commands):
    c = commands[i] #使用字典查詢選字
    #使用迴圈if
    if c == "U":  #為str故要加" "
        y = y + 1
    elif c == "D": #由於題目只有4種結果，故使用elif，不使用else:因為所有字母都會弱在else，包含空白鍵
        y = y - 1
    elif c == "L":
        x = x - 1
    elif c == "R":
        x = x +1
    i = i +1
print(x, y)

#蒙地卡羅算面積
#圓上:x平方 + y平方 = r平方(1)
#圓內:x平方 + y平方 < r平方(1)
import random
count = 0 #起始落到圓裡面
total = 100000 #總共要試100萬次
i = 0 #標準公式 i = 0...
while i < total:
    #對於這次隨機(x, y)
    x = random.uniform(-1, 1)
    y = random.uniform(-1, 1)
    #如果這座標在圓裡面，累積次數增加1
    if x ** 2 + y ** 2 <= 1:
        count = count + 1 #落到圓裡面的標記數目
    i = i + 1
#圓形裡面 / 全部 * 參照面積
ratio = count / total
print(ratio * 4)
```

```python
#四大基礎型態-數字,字串:"hello",布林(Boolean):True/False,無:None
#a = 3, a > 2 True a < 2 False
#要會使用而且(and),或者(or)
import random
win, lose = 0, 0

while win < 3 and lose < 3:
    #-1: lose 0: even 1: win
    result = random.randint(-1, 1)
    print(result)
    if result == -1:
        lose = lose + 1
    elif result == 1:
        win = win + 1
if win > lose:
    print("WIN")
else:
    print("LOSE")
```
