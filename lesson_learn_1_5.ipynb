```python
# bmi: weight(kg) / height(m) ^2
weight = float(input("請輸入你的體重："))
height = float(input("請輸入你的身高："))
bmi = weight / (height / 100) ** 2
print("你的體重是："+ str(weight))
print("你的身高是："+ str(height))
print("bmi是：" + str(bmi))

# >, >=, <, <=, ==
if bmi > 25:
    print("過重")
    print("少吃點")
else:
    print("正常")
```

```python
import random

my = int(input("請輸入0.剪刀 1.石頭 2.布:"))
com = random.randint(0, 2)
trans = ["剪刀", "石頭", "布"]
print("我出的：", trans[my])
print("電腦出的:", trans[com])

if my == (com + 1) % 3:
    print("我贏")
elif com == (my + 1) % 3:
    print("我輸")
else:
    print("平手")
```

```python
"hello" * 3
[1, 2, 3] * 5
i = 0
while i < 10:
    print("*" *(i + 1))
    i = i + 1

left, right = " ", "o"
layer = 10
i = 0
while i < layer:
    lcount = layer - i - 1
    rcount = i + 1
    print(left * lcount + right * rcount)
    i = i + 1

#****
# ***
#  **
#   *

left, right = " ", "x"
layer = 10
i = 0
while i < layer:
    lcount = i
    rcount = layer - i
    print(left * lcount + right * rcount)
    i = i + 1

#*****
# ***
#  *

# 3層: 5
# 4 : 7
# 5 : 9
# layer * 2 - 1

left, mid, right = " ", "c", " "
layer = 10
i = 0
while i < layer:
    lcount = i
    mcount = (layer * 2 - 1) - 2 * i
    rcount = i
    print(left * lcount +
          mid * mcount +
          right * rcount)
    i = i + 1
```
